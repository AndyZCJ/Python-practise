```python
tasks = ["学习Python", "写作业", "锻炼"]
tasks.append("跑步")
if len(tasks) > 1:
    tasks.pop(1)
    print("删除了第二个任务")
else:
    print("列表不足两个元素，无法删除任务")
print("删除后")

if tasks:
    tasks[0] = "深入学习Python"
    print("修改后tasks：", tasks)
else:
    print("任务列表为空")

print(f"未完成的任务数量: {len(tasks)}")
```
