```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
v1 = np.array([2,1])
v2 = np.array([1,3])

v3 = v1 + v2

plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, v3[0], v3[1], angles='xy', scale_units='xy', scale=1)

plt.xlim(0,5)
plt.ylim(0,5)
plt.grid()
plt.show()
```

```python
import numpy as np

rng = np.random.default_rng(0)
a = rng.normal(size=5)
b = rng.normal(size=5)

dot = lambda x,y: float(np.dot(x,y))
norm = lambda x: float(np.linalg.norm(x))

# projection of b onto a
b_par = (dot(a,b)/dot(a,a)) * a
b_perp = b - b_par

print("dot(a,b)     =", dot(a,b))
print("dot(a,b_par) =", dot(a,b_par))
print("dot(a,b_perp)=", dot(a,b_perp))

print("check orthogonality (should be ~0):", dot(a,b_perp))
print("signed proj length of b onto a:", dot(a,b)/norm(a))
print("proj length magnitude:", norm(b_par), " (should equal abs of signed proj length)")
```
